```python
#Import the necessary libraries
import tensorflow as tf
from tensorflow import keras
from keras import layers
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.utils import shuffle
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
```

```python
#Read the Disaster on Social Media dataset using new file paths
train_file_path = '/data/train.csv'
test_file_path = '/data/test_with_label.csv'

train = pd.read_csv(train_file_path)
test = pd.read_csv(test_file_path)

print("Train data head:")
display(train.head())
print("\nTest data head:")
display(test.head())
```

```python
SEED=4242
TARGET = "target"
```

```python
#Shuffle the DataFrame
train_dataset=train
train_dataset = shuffle(train_dataset, random_state=SEED)
train_dataset = shuffle(train_dataset, random_state=int(SEED/2))
train_dataset.head()
```

```python

combined_df = pd.concat([train, test], ignore_index=True)

print("Combined DataFrame shape:", combined_df.shape)
print("\nCombined DataFrame head:")
display(combined_df.head())

print("\nCombined DataFrame tail:")
display(combined_df.tail())
```

```python
from sklearn.model_selection import train_test_split


X_combined = combined_df['text']
y_combined = combined_df['target']


X_train_stratified, X_test_stratified, y_train_stratified, y_test_stratified = train_test_split(
    X_combined, y_combined, test_size=0.2, random_state=42, stratify=y_combined
)

print("Shape of X_train_stratified:", X_train_stratified.shape)
print("Shape of X_test_stratified:", X_test_stratified.shape)
print("Shape of y_train_stratified:", y_train_stratified.shape)
print("Shape of y_test_stratified:", y_test_stratified.shape)

print("\nLabel distribution in y_train_stratified:")
display(y_train_stratified.value_counts(normalize=True))

print("\nLabel distribution in y_test_stratified:")
display(y_test_stratified.value_counts(normalize=True))
```

```python
from sklearn.model_selection import train_test_split

X_train = X_train_stratified
y_train = y_train_stratified

X_test = X_test_stratified
y_test = y_test_stratified
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score


vectorizer = TfidfVectorizer(max_features=15000, ngram_range=(1,2))


X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)


clf = LogisticRegression(max_iter=5000, class_weight='balanced', solver="liblinear")


clf.fit(X_train, y_train)


train_preds = clf.predict(X_train)
test_preds = clf.predict(X_test)
prob_train_stratified = clf.predict_proba(X_train)[:,1]
prob_test_stratified = clf.predict_proba(X_test)[:,1]

# === 평가 ===
print("Train Accuracy:", accuracy_score(y_train, train_preds))
print("Test Accuracy:", accuracy_score(y_test, test_preds))

print("\n[Train Classification Report]")
print(classification_report(y_train, train_preds))

print("\n[Test Classification Report]")
print(classification_report(y_test, test_preds))
```
